```python
import cv2

img = cv2.imread("kwiat.jpg")

img_bw = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

#jupyter notebook
#cv2.imshow('Image', img)

#colab
from google.colab.patches import cv2_imshow
cv2_imshow(img)
```

Thresholding

```python
thresholds = [50, 110, 180]
for t in thresholds:
  ret, thresh = cv2.threshold(img_bw, t, 255, cv2.THRESH_BINARY)

  #cv2.imshow(thresh)
  cv2_imshow(thresh)
```

Gamma correction

```python
import numpy as np
def gamma_correction(image, gamma):
    #image = cv2.imread(img_path)
    gamma_corrected = np.array(255 * (image / 255) ** gamma, dtype='uint8')
    cv2_imshow(gamma_corrected)
    return gamma_corrected
```

```python
import matplotlib.pyplot as plt

gammas = [0.3, 2.2, 10.0]

for g in gammas:
  gamma_corrected = gamma_correction(img, g)
```

Histogram equalization

```python
cv2_imshow(img_bw)
equ = cv2.equalizeHist(img_bw)
cv2_imshow(equ)
```

Median filter

```python
from skimage.morphology import disk
from skimage.filters import median

for i in [3, 5, 11]:
  med = median(img_bw, disk(i))
  cv2_imshow(med)
```

Gaussian filter

```python
cv2_imshow(img)
for s in [5, 11, 19]:
  gau_img1 = cv2.GaussianBlur(img, (s,s), 0) # 5*5 kernal, 2 on each side. 2 = 1/5 * 10 = 1/5 * sigma
  cv2_imshow(gau_img1)
```

Sobel filter

```python
from skimage.filters import sobel

cv2_imshow(img_bw)
for k in [3, 5, 15]:
  filtered = cv2.Sobel(img_bw,cv2.CV_64F,0,1,ksize=k)
  cv2_imshow(filtered)
```

Laplacian filter

```python
cv2_imshow(img_bw)
filtered = cv2.Laplacian(img_bw,cv2.CV_64F)
cv2_imshow(filtered)
```

Morphological tranformations

```python
cloud = cv2.imread("cloud.jpg")
```

Dilation

```python
cv2_imshow(cloud)
kernel = np.ones((5,5),np.uint8)
dilation = cv2.dilate(cloud,kernel,iterations = 1)
cv2_imshow(dilation)
```

Erosion

```python
cv2_imshow(cloud)
kernel = np.ones((5,5),np.uint8)
erosion = cv2.erode(cloud,kernel,iterations = 1)
cv2_imshow(erosion)
```

Opening

```python
cv2_imshow(cloud)
kernel = np.ones((5,5),np.uint8)
opening = cv2.morphologyEx(cloud, cv2.MORPH_OPEN, kernel)
cv2_imshow(opening)
```

Closing

```python
cv2_imshow(cloud)
kernel = np.ones((5,5),np.uint8)
closing = cv2.morphologyEx(cloud, cv2.MORPH_CLOSE, kernel)
cv2_imshow(closing)
```
